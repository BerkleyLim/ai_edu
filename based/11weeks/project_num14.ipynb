## 프로젝트 연구 배경

- 제품 결함 없이 제조 비용을 절감하기 위해 자동화된 품질 관리가 필수적이다.
- 수작업 검사는 생산 증가와 제품 다양화에 따라 비효율적이며 오류 위험이 크다.
- Industry 4.0과 IIoT 발전으로 실시간 공정 모니터링과 대규모 센서 데이터 수집이 가능해졌다.
- 딥러닝 및 컴퓨터 비전 기술은 비침습적이고 신속한 품질 평가를 가능하게 한다.
- 데이터 부족 문제를 해결하기 위해 본 연구는 Siamese 네트워크 기반 One-shot 인식 방법을 제안한다.


`기술 동향`
- 최근 딥러닝과 컴퓨터 비전 기술은 복잡한 이미지 인식 문제를 높은 정확도로 해결하면서, 비침습적 검사 방식의 신속한 품질 평가를 가능하게 하고 있다.
- 딥러닝은 자동으로 특성을 추출하여 전통적 머신러닝보다 뛰어난 인식 성능을 제공하며, 다양한 분야에 적용되고 있다.
-  이러한 기술 발전은 제조 공정의 실시간 모니터링과 품질 관리를 위한 자동화 시스템 구축에 큰 역할을 하고 있다.
- 다만, 기존 기법은 대규모 학습 데이터에 의존하므로, 새로운 제품이나 결함 유형 등장 시 데이터 수집 및 주석 작업이 큰 부담으로 작용한다.
- 이 문제를 극복하기 위해 One-shot, Few-shot 학습 및 전이 학습과 같은 소량의 데이터로도 효과적인 학습 방법에 대한 연구가 주목받고 있다.

## One-Shot Learning 장점

- 소량의 학습 데이터, 나아가 단 한 개의 샘플만으로도 효과적인 학습이 가능하여 데이터 수집 및 라벨링 비용을 크게 절감할 수 있다.
- 신속한 모델 업데이트와 적응이 가능하여, 새로운 제품이나 결함 유형이 등장할 경우에도 빠르게 대응할 수 있다.

## Siamese Network를 사용하였을 때의 기대 효과

- 두 이미지 간의 유사성과 차이를 효과적으로 학습함으로써, 단일 샘플을 기준으로 결함 여부를 정밀하게 판단할 수 있다.
- 실시간 추론이 가능하여 제조 현장에서 자동화된 품질 관리 시스템 구축에 큰 기여를 할 것으로 기대된다.

## 데이터셋 소개

본 연구에서는 Northeastern University (NEU) 표면 결함 데이터베이스를 활용하였다. 동 데이터셋은 열간 압연 강판 표면의 결함 검출을 목적으로 구축된 자료로서, 실제 제조 공정에서 발생 가능한 다종의 결함 유형을 포함하고 있다. 구체적으로, 데이터셋은 하기 여섯 가지 결함 유형으로 구성된다.

- Rolled-in Scale (RS): 압연 과정에서 발생하는 산화 피막이 표면에 불규칙적으로 침투하여 형성되는 결함
- Patches (Pa): 표면에 반점 또는 불규칙한 색상 변화가 발생하는 결함
- Crazing (Cr): 강판 표면에 미세한 균열이 발생하는 결함
- Pitted Surface (PS): 소규모 함몰 또는 구멍이 형성되어 표면이 불균일해지는 결함
- Inclusion (In): 제조 과정 중 금속 내부에 불순물이 포함되어 발생하는 결함
- Scratches (Sc): 외부 충격 또는 마찰에 의해 표면에 긁힘 또는 흠집이 발생하는 결함

데이터셋은 각 결함 클래스당 300매의 그레이스케일 이미지로 구성되어 총 1,800매의 이미지를 포함한다. 각 이미지는 200 × 200 픽셀의 해상도를 가지며, 다양한 조명 조건 및 결함 패턴으로 인해 동일 클래스 내에서도 상당한 변동성을 나타낸다. 이러한 특징은 실제 제조 환경의 복잡성을 반영하는 것으로, 자동화된 품질 관리 시스템 및 결함 인식 알고리즘의 성능 평가에 유효한 자료로 활용될 수 있음을 시사한다.

```python
args = {'training_dataset': '/home/elicer/dataset_all/knn_data/training',
        'testing_dataset': '/home/elicer/dataset_all/knn_data/testing',
        'neighbors': 1,
        'jobs': 1}
```

- 이 셀에서는 훈련 데이터셋과 테스트 데이터셋의 경로, 그리고 KNN 분류기의 동작에 필요한 인자들을 딕셔너리 형태의 변수 args에 정의한다.

```python
def image_to_feature_vector(image, size=(32, 32)):
    return cv2.resize(image, size).flatten()
```

- 이 셀에서는 입력 이미지를 지정된 크기(기본값 32×32)로 리사이즈한 후 1차원 배열로 평탄화하여 반환하는 image_to_feature_vector 함수를 정의한다.
- 이 함수는 원본 이미지의 픽셀 정보를 간단한 벡터 형태로 변환하여 후속 처리에 용이하게 한다.

```python
def extract_color_histogram(image, bins=(8, 8, 8)):
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    hist = cv2.calcHist([hsv], [0, 1, 2], None, bins, [0, 180, 0, 256, 0, 256])
    cv2.normalize(hist, hist)
    return hist.flatten()
```

- 이 셀에서는 입력 이미지의 색상 분포를 나타내는 3차원 히스토그램을 HSV 색상 공간에서 계산하는 extract_color_histogram 함수를 정의한다.
- 이미지를 BGR 색상 공간에서 HSV 색상 공간으로 변환한 후 지정된 bin 수에 따라 히스토그램을 계산하고 정규화한다.
- 이 히스토그램을 1차원 배열로 평탄화하여 특징 벡터로 반환한다.

```python
imagePaths = list(paths.list_images(args["training_dataset"]))
imagePaths_test = list(paths.list_images(args["testing_dataset"]))
```

- 이 셀에서는 imutils.paths.list_images 함수를 사용하여 훈련 데이터셋 및 테스트 데이터셋 폴더 내의 모든 이미지 파일 경로를 탐색한다.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
import random
import torch
import torchvision
import torchvision.transforms as transforms
import torchvision.datasets as datasets
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.autograd import Variable
from torch.utils.data import DataLoader
```

- 운영체제 관련 기능을 위한 os, 수치연산을 위한 numpy, 시각화를 위한 matplotlib.pyplot, 이미지 입출력을 위한 PIL(Image 및 ImageOps), 무작위성을 위한 random 라이브러리이다.
- PyTorch와 관련 모듈들(예: torch, torchvision, torch.nn, torch.optim, torch.nn.functional, torch.autograd, torch.utils.data)이 포함되어 있으며, 이 라이브러리들은 데이터 전처리, 모델 구성, 학습, 평가에 사용된다.

```python
TRAIN_DIR = '/home/elicer/dataset_all/cnn_data/training/'
TEST_DIR = '/home/elicer/dataset_all/cnn_data/testing/'

WEIGHT_PATH = './weights/steel/cnn/'
SAVE_IMAGE_PATH = './images/steel/cnn/'

SAVE_TEST_RESULTS_PATH = './images/steel/test_output'

device = torch.device('cuda:0' if (torch.cuda.is_available()) else 'cpu')

BATCH_SIZE = 128
N_EPOCHS = 80
LR = 0.0005
N_GPU = 1
SET_L = False
INPUT_D = 3 if SET_L == False else 1
```

- 이 셀은 학습 및 평가에 필요한 경로와 하이퍼파라미터들을 정의한다.
- TRAIN_DIR과 TEST_DIR은 각각 훈련 및 테스트 이미지가 저장된 디렉토리 경로를 지정하며, WEIGHT_PATH, SAVE_IMAGE_PATH, SAVE_TEST_RESULTS_PATH는 모델 파라미터와 결과 이미지를 저장할 경로를 설정한다.
- 또한, device 변수는 GPU가 사용 가능할 경우 'cuda:0'를, 그렇지 않으면 'cpu'를 사용하도록 설정하며, 배치 크기, 에폭 수, 학습률 및 입력 채널 수 등 모델 학습에 필요한 기타 설정들도 함께 정의되어 있다.

```python
class CNNNet(nn.Module):
    def __init__(self):
        super(CNNNet, self).__init__()
        self.conv_layer = nn.Sequential(
            nn.ZeroPad2d(1),
            nn.Conv2d(INPUT_D, 4, kernel_size=3),
            nn.ReLU(),        
            nn.ZeroPad2d(1),
            nn.Conv2d(4, 8, kernel_size=3),
            nn.ReLU(),
            nn.ZeroPad2d(1),
            nn.Conv2d(8, 8, kernel_size=3),
            nn.ReLU(),
        )
        
        self.fc_layer = nn.Sequential(
            nn.Linear(8*100*100, 500),
            nn.ReLU(),
            nn.Linear(500, 500),
            nn.ReLU(),
            nn.Linear(500, 6),
        )
        
    def forward(self, x):
        x = self.conv_layer(x)
        x = x.view(x.size()[0], -1)
        x = self.fc_layer(x)
        return x
```

- 이 셀은 CNN 기반 분류 모델인 CNNNet 클래스를 정의한다.
- 모델은 먼저 convolutional 레이어들을 이용하여 입력 이미지에서 공간적 특징을 추출하는 conv_layer로 구성되며, 그 후 전결합층(fc_layer)을 통해 추출된 특징을 6개의 클래스에 대응하는 출력으로 변환한다.
- 입력 이미지의 채널 수는 INPUT_D로 결정되며, 이미지 크기가 100×100으로 가정되어 Fully Connected 레이어의 입력 차원(8×100×100)이 설정된다.

```python
def show_img(img, text=None, figsize=(20,20), save_into=None):
    img = img.numpy()
    plt.figure(figsize=figsize)
    plt.axis('off')
    if text:
        plt.text(50, 8, text, bbox={'facecolor': 'white', 'alpha': 1 })
    plt.imshow(np.transpose(img, (1, 2, 0)))
    
    if save_into:
        plt.savefig(save_into)
        plt.show()
    else:
        plt.show()
```

- 이 셀은 이미지 또는 이미지 그리드를 시각화하고 저장하는 show_img 함수를 정의한다.
- 함수는 입력으로 받은 이미지 텐서를 NumPy 배열로 변환한 후, 배열의 차원 순서를 변경하여 올바른 형식(RGB 순)으로 이미지를 표시한다.
- 옵션으로 텍스트를 삽입할 수 있으며, 지정된 파일 경로가 있을 경우 해당 경로에 이미지를 저장한 후 출력한다.

```python
def explain_databatch(batch):
    """
    batch: tensor batch from dataset
    """
    print('Each batch of training data is a tuple of {} elements.'.format(len(batch)))
    
    print('Shape of each element of a tuple in the batch:')
    for i in range(len(batch)):
        print('element#{}: {}'.format(i, batch[i].shape))
```

- 이 셀은 주어진 각도들 중 하나를 무작위로 선택하여 이미지를 회전시키는 RandRotateTransform 클래스를 정의한다.
- 객체가 호출될 때마다 angles 목록에서 임의의 각도를 선택하고, 해당 각도로 입력 이미지를 회전시킨 결과를 반환한다.

```python
class RandRotateTransform:
    """Rotate by one of the given angles."""

    def __init__(self, angles):
        self.angles = angles

    def __call__(self, x):
        angle = random.choice(self.angles)
        return transforms.functional.rotate(x, angle)
```

- 이 셀은 RandRotateTransform 클래스는 지정된 각도 목록 중 하나를 무작위로 선택하여 입력 이미지를 회전시키는 데이터 증강 기능을 제공한다.
- 생성자에서는 전달받은 각도 리스트를 객체 변수에 저장하며, 호출 시 random.choice 함수를 사용해 임의의 회전 각도를 결정한다.
- 선택된 각도는 transforms.functional.rotate 함수에 적용되어 회전된 이미지를 반환하며, 이를 통해 모델 학습 시 데이터 다양성이 향상된다.


```python
transform = transforms.Compose([
                            transforms.Resize((100,100)),
                            RandRotateTransform(angles=[0, 90, 180, 270]),
                            transforms.RandomHorizontalFlip(),
                            transforms.ToTensor(),
                            transforms.Normalize((0.5,), (0.5,))
])
```

- 이 셀은 이미지 전처리를 위한 파이프라인 transform을 구성한다.
- 파이프라인은 이미지를 100×100 크기로 리사이즈하고, RandRotateTransform을 통해 지정된 각도(0, 90, 180, 270도) 중 하나로 회전시킨 후, 랜덤 수평 반전을 적용한다.
- 그 후 이미지를 텐서로 변환하고 픽셀 값을 정규화하여 모델 입력에 적합한 형태로 만든다.

```python
trainset = torchvision.datasets.ImageFolder(root=TRAIN_DIR, transform=transform)
train_loader = torch.utils.data.DataLoader(trainset, batch_size=BATCH_SIZE,
                                          shuffle=True, num_workers=0)

example_batch = next(iter(train_loader))

explain_databatch(example_batch)
```

- 이 셀은 ImageFolder를 사용하여 지정된 경로(TRAIN_DIR)의 훈련 데이터셋을 로드하고, 구성한 transform을 적용한다.
- 이후, DataLoader를 사용하여 데이터를 배치 단위로 로드하며, 배치 크기는 BATCH_SIZE로 설정된다.
- 데이터 로더에서 한 배치를 추출하여 explain_databatch 함수를 호출함으로써, 배치 데이터의 구조와 각 요소의 텐서 차원을 출력하여 확인한다.

```python
model = CNNNet().to(device)
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=LR)
```

- 이 셀은 CNNNet 클래스로부터 CNN 모델을 생성하고, 지정된 장치(device)로 모델을 이동시킨다.
- 또한, 다중 클래스 분류 문제에 적합한 손실 함수로 CrossEntropyLoss를, 모델의 파라미터 업데이트를 위해 Adam 최적화 알고리즘을 설정힌다.

```python
for epoch in range(120):
    running_loss = 0.0
    for i, data in enumerate(train_loader, 0):
        inputs, labels = data
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()

        # forward + backward + optimize
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        
        if i % 10 == 0:
            running_loss = 0.0
    
    print('[%d] \t loss: %.6f' % (epoch + 1, running_loss / 2000))

print('\n\nFinished Training!!')
```


- 이 셀은 120 에폭 동안 모델의 학습을 수행하는 루프를 실행한다.
- 각 배치마다, 입력 데이터와 레이블을 지정된 장치로 이동시키고 optimizer의 기울기를 초기화한 다음, 모델의 forward pass를 통해 출력을 얻는다.
- 이후, 손실 값을 계산하고 역전파(backward)를 통해 기울기를 구한 후 optimizer를 통해 파라미터를 업데이트힌다.
- 일정 간격(10 배치마다)마다 누적 손실을 초기화하고 각 에폭이 끝날 때마다 평균 손실 값을 출력하여 학습 상태를 모니터링하며, 최종적으로 학습이 완료되었음을 출력한다.

```python
def explain_databatch(batch):
    """
    batch: tensor batch from dataset
    """
    print('Each batch of training data is a tuple of {} elements.'.format(len(batch)))
    
    print('Shape of each element of a tuple in the batch:')
    for i in range(len(batch)):
        print('element#{}: {}'.format(i, batch[i].shape))
```

- 함수는 입력으로 받은 배치가 튜플로 구성되어 있음을 출력하고, 튜플 내 각 요소의 텐서 형태(차원)를 순차적으로 출력하여 데이터 배치의 구조를 명확하게 파악할 수 있도록 한다.

```python
class SiameseDataset(torch.utils.data.Dataset):
    
    def __init__(self,
                 datasets, 
                 train=True,
                 transform=None,
                 should_invert=True,
                 set_luminance=True):
        
        self.datasets = datasets
        self.train = train
        self.transform = transform
        self.should_invert = should_invert
        self.set_luminance = set_luminance
        
    def __getitem__(self, index):
        
        imageA_tupple = random.choice(self.datasets.imgs)
        get_same_class = random.randint(0,1)
        if get_same_class:
            while True:
                imageB_tupple = random.choice(self.datasets.imgs)
                if imageA_tupple[1] == imageB_tupple[1]:
                    break
        else:
            while True:
                imageB_tupple = random.choice(self.datasets.imgs)
                if imageA_tupple[1] != imageB_tupple[1]:
                    break
                    
        imageA = Image.open(imageA_tupple[0])
        imageB = Image.open(imageB_tupple[0])
        
        if self.set_luminance:
            imageA = imageA.convert('L')
            imageB = imageB.convert('L')
        
        if self.should_invert:
            imageA = ImageOps.invert(imageA)
            imageB = ImageOps.invert(imageB)
            
        if self.transform is not None:
            imageA = self.transform(imageA)
            imageB = self.transform(imageB)
        
        if self.train:
            return imageA, imageB, torch.from_numpy(np.array([int(imageA_tupple[1] != imageB_tupple[1])],
                                              dtype=np.float32))
        else:
            return imageA, imageA_tupple[0], imageA_tupple[1], imageB, imageB_tupple[0], imageB_tupple[1]
        
    def __len__(self):
        return len(self.datasets.imgs)
```

- 이 셀은 Siamese 네트워크 학습을 위한 데이터셋 클래스를 정의하는 SiameseDataset을 구현한다.
- 생성자에서는 전달받은 datasets (ImageFolder 객체), 전처리 transform, 그리고 이미지 반전 여부와 밝기 설정을 저장한다.
- __getitem__ 메서드는 무작위로 두 개의 이미지 쌍을 선택하여, 동일 클래스 혹은 서로 다른 클래스의 쌍을 생성한다.
- 이후 각 이미지의 밝기 변환 및 반전 옵션을 적용하고, 지정된 전처리 transform을 통해 텐서로 변환한 후, 학습 모드에서는 두 이미지와 두 이미지의 레이블(동일/비동일)을 반환한다.
- __len__은 데이터셋 내 이미지 수를 반환한다.

```python
class RandRotateTransform:
    """Rotate by one of the given angles."""

    def __init__(self, angles):
        self.angles = angles

    def __call__(self, x):
        angle = random.choice(self.angles)
        return transforms.functional.rotate(x, angle)
```

- 이 셀은 입력 이미지에 대해 주어진 각도 리스트 중 하나를 무작위로 선택하여 이미지를 회전시키는 RandRotateTransform 클래스를 정의한다.
- 객체가 호출되면, angles에서 임의의 각도를 선택하여 회전된 이미지를 반환한다.

```python
transform = transforms.Compose([
                            transforms.Resize((100,100)),
                            RandRotateTransform(angles=[0, 90, 180, 270]),
                            transforms.RandomHorizontalFlip(),
                            transforms.ToTensor(),
                            transforms.Normalize((0.5,), (0.5,))
])
```

- 이후, 시각화를 위한 예시 배치를 추출한 후 explain_databatch를 통해 배치 구조를 다시 한 번 확인하고, 일부 이미지들을 결합하여 그리드 형태로 시각화 및 저장한다.
- 마지막으로, 결합된 이미지 그리드의 텐서 형태를 출력한다.

```python
class SiamaseNet(nn.Module):
    
    def __init__(self):
        super(SiamaseNet, self).__init__()
        self.conv_layer = nn.Sequential(
            nn.ZeroPad2d(1),
            nn.Conv2d(INPUT_D, 4, kernel_size=3),
            nn.ReLU(),        
            nn.ZeroPad2d(1),
            nn.Conv2d(4, 8, kernel_size=3),
            nn.ReLU(),
            nn.ZeroPad2d(1),
            nn.Conv2d(8, 8, kernel_size=3),
            nn.ReLU(),
        )
        
        self.fc_layer = nn.Sequential(
            nn.Linear(8*100*100, 500),
            nn.ReLU(),
            nn.Linear(500, 500),
            nn.ReLU(),
            nn.Linear(500, 5),
        )
        
    def forward_once(self, x):
        x = self.conv_layer(x)
        x = x.view(x.size()[0], -1)
        x = self.fc_layer(x)
        
        return x
    
    def forward(self, imageA, imageB):
        resultA = self.forward_once(imageA)
        resultB = self.forward_once(imageB)
        
        return resultA, resultB
```

- 이 셀에서는 시암 네트워크 기반의 모델인 SiamaseNet 클래스를 정의한다.
- 본 모델은 두 이미지에 대해 동일한 convolutional 레이어들을 사용하여 각 이미지의 특징을 추출하고, Fully Connected 레이어를 통해 5차원 출력 벡터로 변환한다.
- forward_once 메서드는 단일 이미지에 대해 레이어들을 순차적으로 적용하는 기능을 수행하며, forward 메서드는 두 이미지(예: 이미지 A와 이미지 B)를 각각 처리하여 그 결과를 반환한다.

```python
siamese = SiamaseNet().to(device)
```

- 이 셀에서는 SiamaseNet 모델을 인스턴스화한 후, 지정된 연산 장치(device)로 모델을 이동시킨다.

```python
class ContrastiveLoss(torch.nn.Module):
    
    def __init__(self, margin=2.0):
        super(ContrastiveLoss, self).__init__()
        self.margin = margin
        
    def forward(self, resultA, resultB, label):
        euclidean_distance = 0.5*F.pairwise_distance(resultA, resultB, keepdim=True) # L2 Norm
        
        contrastive_loss = torch.mean((1-label) * torch.pow(euclidean_distance, 2) + 
                        (label) * torch.pow(torch.clamp(self.margin - euclidean_distance, min=0.0), 2))
        
        return contrastive_loss
```

- 이 셀에서는 대조적 손실(Contrastive Loss)을 구현한 ContrastiveLoss 클래스를 정의한다.
- 손실 함수는 두 입력 이미지의 특징 벡터 간의 유클리드 거리를 기반으로 동일 클래스인지 여부에 따라 벌점을 부여하며, margin 값을 이용해 서로 다른 클래스 간 거리가 일정 수준 이상 유지되도록 조정한다.

```python
contrastive_loss = ContrastiveLoss()
optimizer = optim.Adam(siamese.parameters(), lr=LR)

train_loss = 0
train_loss_history = []

N = len(datasets)

for epoch in range(N_EPOCHS):
    for batch_idx, data in enumerate(train_loader):
        imageA, imageB, label = data
        imageA, imageB, label = imageA.to(device), imageB.to(device), label.to(device)
        
        optimizer.zero_grad()
        resultA, resultB = siamese(imageA, imageB)
        loss = contrastive_loss(resultA, resultB, label)
        
        loss.backward()
        optimizer.step()
        
        train_loss += loss.item()
    
    BatchCount_PER_EPOCH = N//BATCH_SIZE
    train_loss = train_loss/BatchCount_PER_EPOCH
    train_loss_history.append(train_loss)
    
    print('Train Epoch:{}\t\tLoss:{:.6f}'.format(epoch+1, train_loss))

    if not os.path.exists(WEIGHT_PATH):
        os.makedirs(WEIGHT_PATH)
    torch.save(siamese, os.path.join(WEIGHT_PATH, 'siamese_{:03}.pt'.format(epoch+1)))
    train_loss = 0
    
    siamese.train()
```


- 이 셀은 시암 네트워크 모델을 대조적 손실 함수를 이용해 학습하는 루프를 수행한다.
- 각 에폭마다 데이터 로더로부터 배치를 추출하여, 이미지 A와 B를 지정된 장치로 이동시킨 후, 모델의 forward pass를 통해 두 이미지의 특징 벡터를 얻는다.
- 이어서 대조적 손실을 계산하고 역전파 및 최적화 단계를 통해 모델 파라미터를 업데이트한다.
- 에폭마다 평균 손실을 계산하여 출력하고, 가중치 저장 경로가 없으면 생성한 후 각 에폭이 끝날 때마다 모델을 저장한다.

```python
SAVE_TEST_RESULTS_PATH = "./assets/paper"
SAVE_TEST_RESULTS = "./assets/paper"

test_loader = DataLoader(siamese_test_datasets,
                         num_workers=0,
                         batch_size=1,
                         shuffle=True)

data_iter = iter(test_loader)

correct_ = 0
count = 0
for i in range(len(siamese_test_datasets)):
    try:
        count += 1
        imageA, pathA, classA, imageB, pathB, classB = next(data_iter)
    except:
        break
    
    with torch.no_grad():
        resultA, resultB = siamese(Variable(imageA).to(device), Variable(imageB).to(device))
        euclidean_distance = F.pairwise_distance(resultA, resultB)
        
        if euclidean_distance >= 2. and classA != classB:
            correct_ += 1
        elif euclidean_distance < 2. and classA == classB:
            correct_ += 1
                
    concatenated = torch.cat((imageA, imageB), 0)
    
    # Uncomment below lines to save figures
    image_file_name = 'test_output_{}.png'.format(i)
    save_results_in = os.path.join(SAVE_TEST_RESULTS_PATH, image_file_name)
```

- 이 셀은 테스트 데이터셋 전체에 대해 시암 네트워크 모델의 성능을 평가하는 반복문을 실행한다.
- 각 반복마다 배치에서 두 이미지(이미지 A와 B)를 추출하고, 두 이미지가 같은 클래스인지 여부를 결정하기 위해 대조적 손실 기준으로 두 특징 벡터 간의 유클리드 거리를 계산한다.
- 만약 거리가 margin 값(여기서는 2.0) 이상이고 두 이미지가 서로 다른 클래스라면, 또는 거리가 margin 미만이고 두 이미지가 같은 클래스라면 정답으로 간주하여 정답 수를 누적한다.
- 각 배치에서 두 이미지를 결합하여 시각화용 그리드를 생성하며, 이후 지정된 경로에 결과 이미지를 저장할 수 있도록 파일 이름과 저장 경로를 구성한다.

```python
accuracy = correct_ / count * 100
print(f"One-shot Siamese accuracy: {accuracy:.2f}%")
```

- 해당 두 줄 코드는 테스트 루프에서 계산한 정답 개수(correct_)와 전체 샘플 개수(count)를 이용해 정확도(accuracy) 를 백분율 형태로 계산하고, 그 결과를 소수점 둘째 자리까지 포맷해 출력하는 역할을 한다.
